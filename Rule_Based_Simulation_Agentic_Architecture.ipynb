```python
%%writefile agentic_gtm_orchestrator.py
import json
import pandas as pd
from datetime import datetime
from typing import List, Dict

# ==========================================
# AGENT 1: UNSTRUCTURED SIGNAL INGESTION AGENT
# ==========================================
class SentimentSignalAgent:
    """Simulates ingestion and triage of unstructured public customer feedback & service friction logs."""

    def __init__(self, target_account: str):
        self.target_account = target_account

    def fetch_unstructured_signals(self) -> pd.DataFrame:
        raw_signals = [
            {"source": "X/Twitter", "category": "Mobile App Outage", "volume_spike": "+340%", "sentiment_score": -0.82, "urgency": "High", "text": "App down again during end-of-month payroll processing. Third time this quarter!"},
            {"source": "HelloPeter", "category": "Conduct Risk / TCF", "volume_spike": "+120%", "sentiment_score": -0.75, "urgency": "High", "text": "Unresponsive dispute resolution team regarding unauthorized debit orders."},
            {"source": "News24", "category": "Regulatory Scrutiny", "volume_spike": "N/A", "sentiment_score": -0.60, "urgency": "Medium", "text": "Regulators signal tighter enforcement on customer dispute turnaround times for tier-1 banks."},
            {"source": "LinkedIn", "category": "Competitor Switch", "volume_spike": "+45%", "sentiment_score": -0.55, "urgency": "Medium", "text": "Considering switching business accounts due to lack of real-time fraud alerts."}
        ]
        return pd.DataFrame(raw_signals)

# ==========================================
# AGENT 2: MEDDPICC RISK & PAIN MAPPER AGENT
# ==========================================
class MEDDPICCMappingAgent:
    """Maps operational sentiment signals directly to MEDDPICC enterprise sales frameworks."""

    def analyze_signals(self, signals_df: pd.DataFrame) -> Dict[str, str]:
        high_urgency = signals_df[signals_df['urgency'] == 'High']

        meddpicc_analysis = {
            "Implicated_Buyer_Personas": "Chief Risk Officer (CRO), Chief Customer Experience Officer (CCXO), Head of Conduct Risk",
            "Identified_Pain": f"Severe customer churn risk caused by {high_urgency['category'].iloc[0]} and regulatory audit exposure under TCF frameworks.",
            "Economic_Metrics": "Estimated 2.4% quarterly churn risk on retail accounts; potential regulatory fines for delayed dispute resolutions.",
            "Decision_Criteria": "Platform must combine real-time AI sentiment analysis with human-in-the-loop verification for compliance reporting.",
            "DataEQ_Value_Driver": "Deploy DataEQ Risk & CX Governance suite to automate early-warning outage detection and TCF compliance audits."
        }
        return meddpicc_analysis

# ==========================================
# AGENT 3: AE EXECUTIVE BRIEFING GENERATOR
# ==========================================
class CommercialBriefingAgent:
    """Synthesizes agent outputs into an AE/ESDM account intelligence briefing."""

    def generate_markdown_brief(self, target_account: str, signals_df: pd.DataFrame, meddpicc: Dict[str, str]) -> str:
        timestamp = datetime.now().strftime("%Y-%m-%d")

        brief = f"# AGENTIC GTM ACCOUNT BRIEF: {target_account.upper()}\n"
        brief += f"**Generated by DataEQ AI-GTM Orchestrator** | *Date: {timestamp}*\n\n"
        brief += "---\n\n"

        brief += "## 1. Unstructured Public Data & Risk Signals\n"
        brief += "Below is the real-time triaged sentiment feed ingested across social, review, and news channels:\n\n"
        brief += signals_df.to_markdown(index=False)
        brief += "\n\n"

        brief += "## 2. MEDDPICC Enterprise Commercial Analysis\n"
        for key, val in meddpicc.items():
            formatted_key = key.replace("_", " ")
            brief += f"- **{formatted_key}:** {val}\n"

        brief += "\n## 3. ESDM / AE Actionable Outreach Strategy\n"
        brief += f"**Target Prospecting Angle for {meddpicc['Implicated_Buyer_Personas'].split(',')[0]}:**\n"
        brief += f"> *\"Our automated monitoring flagged a {signals_df['volume_spike'].iloc[0]} spike in customer friction around mobile platform stability for {target_account}. DataEQ helps banking CROs quantify these operational risks in real-time before they impact TCF compliance scores or drive account churn. Let's schedule 15 minutes to review your quarterly benchmark report.\"*\n"

        return brief

# ==========================================
# PIPELINE EXECUTION ENGINE
# ==========================================
def main():
    target_company = "Standard Bank South Africa"
    print(f"[+] Running Agentic GTM Pipeline for: {target_company}...")

    ingest_agent = SentimentSignalAgent(target_company)
    signals = ingest_agent.fetch_unstructured_signals()

    meddpicc_agent = MEDDPICCMappingAgent()
    meddpicc_insights = meddpicc_agent.analyze_signals(signals)

    brief_agent = CommercialBriefingAgent()
    final_report = brief_agent.generate_markdown_brief(target_company, signals, meddpicc_insights)

    output_filename = "EXECUTIVE_ACCOUNT_BRIEF.md"
    with open(output_filename, "w") as f:
        f.write(final_report)

    print(f"[✓] Pipeline complete. Generated executive intelligence report: {output_filename}")

if __name__ == "__main__":
    main()
```
